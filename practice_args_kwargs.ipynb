## Practice *args and **kwargs

```python
def show_args(*args):
    print(args)

show_args(7,7,8)
```

```python
def show_kwargs(*args, **kwargs):
    print(kwargs)

show_kwargs(a = 6, b = 8)
```

```python
show_kwargs(6, 8)
```

## Undertsanding decorators

```python
def log_and_time(func):
    def wrapper(*args, **kwargs):
        print("before function")
        result = func(*args, **kwargs)
        print("after function")
        return result
    return wrapper
```

```python
def say_hello():
    print("Hello")
```

```python
say_hello = log_and_time(say_hello)
```

```python
say_hello
```

```python
say_hello()
```

```python
@log_and_time
def add(a:int, b:int) -> int:
    return(int(a+b))
```

```python
add(5,3)
```
